```python
# Data Frame

data = {
    "Name": ["Rajan", "Amit", "Priya", "Neha", "Karan"],
    "Age": [22, 21, 22, 20, 23],
    "Marks": [85, 78, 92, 67, 88]
}

df = pd.DataFrame(data)

print(df, type(df))

print(df.index)

print(df.columns)

print(df.shape)

# print(df.describe)

# mean 
print("The Mean of Marks: ", df["Marks"].mean())
```

```python
# Create Data Frame through Numpy

import numpy as np
import pandas as pd

np_arr = np.array([ 
    ["Eve", "Bob"],
    [23, 25]
])

df = pd.DataFrame(np_arr.T, columns=["Name", "Age"])

print(df)
```
