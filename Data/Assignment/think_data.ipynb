# Load the Data 

```python
def load_file(filename):
    with open(filename, "r") as f:
        data = json.load(f)

        return data

```

```python
data = load_file("store_data.json")
print(data)
```

# clean and Structize the data

```python

def clean_data(data):
    
    txt_to_num = {
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5
    }

    cleaned_data = []

    unique_user= set()

    for user in data:

        raw_rating = user["rating"]

        if raw_rating is None:
            print("Missing rating")
            continue

        if isinstance(raw_rating, str):
            raw_rating = raw_rating.strip().lower()

            if raw_rating in txt_to_num:
                raw_rating = txt_to_num[raw_rating]

        user["rating"] = raw_rating

    # Handling missing value

        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

    # Validate Rating
        if (not isinstance(raw_rating, (int ,float)) or isinstance(raw_rating, bool) or not 1 <= raw_rating <= 5):
            print(f"Invalid rating for {user['name']}")
            continue
    

    #Deduplication

        normalize_name = user["name"].strip().lower()



        record_key = (
            normalize_name,
            user.get("rating"),
            user.get("feedback"),
            user.get("age")
        )

        if(record_key in unique_user):
            continue
        unique_user.add(record_key)
        cleaned_data.append(user)

    return cleaned_data


    




```

```python
clean_data(data)
```
